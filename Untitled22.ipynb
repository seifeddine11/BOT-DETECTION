```python

"""
================================================================================
MULTI-LABEL DEEP LEARNING FOR DRUG MECHANISM OF ACTION (MoA) PREDICTION
FROM TRANSCRIPTOMIC AND CELL VIABILITY PROFILES
================================================================================
Pipeline complet : téléchargement, EDA, preprocessing, modélisation,
validation croisée groupée par drug_id, évaluation, visualisations.
================================================================================
"""

# ============================================================================
# 0. INSTALLATION DES DÉPENDANCES
# ============================================================================
!pip install gdown lightgbm -q

# ============================================================================
# 1. IMPORTS
# ============================================================================
import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import wilcoxon
from pathlib import Path

from sklearn.preprocessing import QuantileTransformer, RobustScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import GroupKFold
from sklearn.metrics import log_loss, roc_auc_score, average_precision_score

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader

import lightgbm as lgb
import gdown

warnings.filterwarnings('ignore')

# ============================================================================
# 2. CONFIGURATION GLOBALE
# ============================================================================
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {DEVICE}")

CONFIG = {
    'n_folds': 5,
    'batch_size': 256,
    'epochs': 50,
    'learning_rate': 1e-3,
    'weight_decay': 1e-5,
    'hidden_dims': [1024, 512, 256],
    'dropout': 0.3,
    'patience': 10,
    'output_dir': './results'
}

Path(CONFIG['output_dir']).mkdir(exist_ok=True)


# ============================================================================
# 3. CHARGEMENT DES DONNÉES (avec téléchargement automatique)
# ============================================================================
class MoADataLoader:
    """
    Classe de chargement et fusion des données MoA.
    Télécharge automatiquement les fichiers manquants depuis Google Drive.
    """

    FILE_IDS = {
        'train_features.csv': '14m_wTanWhK9McnKCvbsQbO7hXSTsaLy0',
        'train_targets_scored.csv': '1xeS8ht9yyi_iNtBfv5L2XFlZGSFXNlzM',
        'train_targets_nonscored.csv': '1Li55cDtZyIiA9SS37_ZW3vNccQDtNAaa',
        'train_drug.csv': '1pyrRlbxULkfPp-ANxr6uUHHSYRGHrn0m',
        'test_features.csv': '1CuvWadjCgdm6rV7UhnzsT5thcXhjRhpD'
    }

    def __init__(self, data_dir='./'):
        self.data_dir = Path(data_dir)
        self.gene_cols = None
        self.cell_cols = None
        self.target_cols = None
        self.nonscored_cols = None

    def _ensure_files_exist(self):
        for filename, file_id in self.FILE_IDS.items():
            filepath = self.data_dir / filename
            if not filepath.exists():
                print(f"⬇️ Téléchargement de {filename}...")
                url = f"https://drive.google.com/uc?id={file_id}"
                gdown.download(url, str(filepath), quiet=False)
            else:
                print(f"✅ {filename} déjà présent")

    def load_all(self):
        print("📂 Vérification / téléchargement des données...")
        self._ensure_files_exist()

        print("\n📂 Chargement des données en mémoire...")
        self.train_features = pd.read_csv(self.data_dir / 'train_features.csv')
        self.train_scored = pd.read_csv(self.data_dir / 'train_targets_scored.csv')
        self.train_nonscored = pd.read_csv(self.data_dir / 'train_targets_nonscored.csv')
        self.train_drug = pd.read_csv(self.data_dir / 'train_drug.csv')
        self.test_features = pd.read_csv(self.data_dir / 'test_features.csv')

        self.gene_cols = [c for c in self.train_features.columns if c.startswith('g-')]
        self.cell_cols = [c for c in self.train_features.columns if c.startswith('c-')]
        self.target_cols = [c for c in self.train_scored.columns if c != 'sig_id']
        self.nonscored_cols = [c for c in self.train_nonscored.columns if c != 'sig_id']

        print(f"✅ Features: {len(self.gene_cols)} gènes + {len(self.cell_cols)} cellulaires")
        print(f"✅ Targets scored: {len(self.target_cols)}")
        print(f"✅ Targets nonscored: {len(self.nonscored_cols)}")

        return self

    def merge_train_data(self):
        df = self.train_features.merge(self.train_drug, on='sig_id', how='left')
        df = df.merge(self.train_scored, on='sig_id', how='left')
        self.train_merged = df
        return df

    def get_feature_target_arrays(self, df, filter_ctl=True):
        if filter_ctl:
            df = df[df['cp_type'] == 'trt_cp'].copy()
        X = df[self.gene_cols + self.cell_cols].values
        y = df[self.target_cols].values
        groups = df['drug_id'].values if 'drug_id' in df.columns else None
        return X, y, groups, df


# ============================================================================
# 4. EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================================
class MoAExploratoryAnalysis:
    """Génère les statistiques et visualisations pour l'article scientifique."""

    def __init__(self, loader: MoADataLoader):
        self.loader = loader

    def compute_summary_statistics(self):
        stats_dict = {
            'N_samples_train': len(self.loader.train_features),
            'N_samples_test': len(self.loader.test_features),
            'N_gene_features': len(self.loader.gene_cols),
            'N_cell_features': len(self.loader.cell_cols),
            'N_MoA_scored': len(self.loader.target_cols),
            'N_MoA_nonscored': len(self.loader.nonscored_cols),
            'N_unique_drugs': self.loader.train_drug['drug_id'].nunique(),
            'Avg_replicates_per_drug': len(self.loader.train_drug) /
                                        self.loader.train_drug['drug_id'].nunique(),
            'Pct_ctl_vehicle_train': (self.loader.train_features['cp_type'] ==
                                       'ctl_vehicle').mean() * 100,
            'Mean_MoA_activation_rate': self.loader.train_scored[
                self.loader.target_cols].mean().mean() * 100,
        }
        return pd.DataFrame([stats_dict]).T.rename(columns={0: 'Value'})

    def analyze_label_imbalance(self):
        target_cols = self.loader.target_cols
        df_scored = self.loader.train_scored

        moa_counts = df_scored[target_cols].sum().sort_values(ascending=False)

        imbalance_stats = {
            'MoA_zero_positive': (moa_counts == 0).sum(),
            'MoA_less_10_positive': (moa_counts < 10).sum(),
            'MoA_less_50_positive': (moa_counts < 50).sum(),
            'MoA_more_500_positive': (moa_counts > 500).sum(),
            'Max_MoA_count': moa_counts.max(),
            'Min_MoA_count': moa_counts.min(),
            'Median_MoA_count': moa_counts.median(),
            'Mean_labels_per_sample': df_scored[target_cols].sum(axis=1).mean(),
            'Max_labels_per_sample': df_scored[target_cols].sum(axis=1).max(),
            'Pct_samples_zero_labels': (
                df_scored[target_cols].sum(axis=1) == 0).mean() * 100
        }
        return moa_counts, imbalance_stats

    def analyze_drug_leakage_risk(self):
        drug_counts = self.loader.train_drug['drug_id'].value_counts()
        leakage_stats = {
            'N_unique_drugs': len(drug_counts),
            'Mean_replicates': drug_counts.mean(),
            'Median_replicates': drug_counts.median(),
            'Max_replicates': drug_counts.max(),
            'Drugs_single_sample': (drug_counts == 1).sum(),
            'Drugs_gt20_samples': (drug_counts > 20).sum(),
        }
        return drug_counts, leakage_stats

    def generate_figure_1_overview(self, save_path='figure1_overview.png'):
        fig, axes = plt.subplots(2, 2, figsize=(14, 10))

        target_cols = self.loader.target_cols
        n_active = self.loader.train_scored[target_cols].sum(axis=1)
        axes[0, 0].hist(n_active, bins=range(0, 8), color='#3498db',
                         edgecolor='black', align='left')
        axes[0, 0].set_xlabel('Number of Active MoA Labels')
        axes[0, 0].set_ylabel('Number of Samples')
        axes[0, 0].set_title('A. Label Cardinality Distribution')
        axes[0, 0].set_xticks(range(0, 8))

        moa_counts = self.loader.train_scored[target_cols].sum().sort_values(ascending=False)
        axes[0, 1].plot(range(len(moa_counts)), moa_counts.values,
                         color='#e74c3c', linewidth=2)
        axes[0, 1].set_yscale('log')
        axes[0, 1].set_xlabel('MoA Rank')
        axes[0, 1].set_ylabel('Positive Count (log scale)')
        axes[0, 1].set_title('B. Label Frequency Distribution')
        axes[0, 1].axhline(y=10, color='gray', linestyle='--', alpha=0.5, label='n=10')
        axes[0, 1].legend()

        drug_counts = self.loader.train_drug['drug_id'].value_counts()
        axes[1, 0].hist(drug_counts.values, bins=50, color='#2ecc71', edgecolor='black')
        axes[1, 0].set_xlabel('Number of Replicates per Drug')
        axes[1, 0].set_ylabel('Number of Drugs')
        axes[1, 0].set_title('C. Drug Replication Distribution')
        axes[1, 0].axvline(drug_counts.mean(), color='red', linestyle='--',
                            label=f'Mean={drug_counts.mean():.1f}')
        axes[1, 0].legend()

        train_dist = self.loader.train_features['cp_type'].value_counts(normalize=True)
        test_dist = self.loader.test_features['cp_type'].value_counts(normalize=True)
        comparison = pd.DataFrame({'Train': train_dist, 'Test': test_dist})
        comparison.plot(kind='bar', ax=axes[1, 1], color=['#9b59b6', '#f39c12'])
        axes[1, 1].set_title('D. Treatment Type: Train vs Test')
        axes[1, 1].set_ylabel('Proportion')
        axes[1, 1].set_xticklabels(comparison.index, rotation=0)

        plt.tight_layout()
        plt.savefig(f"{CONFIG['output_dir']}/{save_path}", dpi=300, bbox_inches='tight')
        plt.show()
        return fig


# ============================================================================
# 5. FEATURE ENGINEERING
# ============================================================================
class FeatureEngineer:
    """Pipeline de preprocessing des features (RankGauss + PCA)."""

    def __init__(self, method='rankgauss'):
        self.method = method
        self.scaler = None
        self.pca_gene = None
        self.pca_cell = None

    def fit_transform(self, X_train, gene_idx, cell_idx, n_pca_gene=80, n_pca_cell=10):
        if self.method == 'rankgauss':
            self.scaler = QuantileTransformer(
                n_quantiles=100, output_distribution='normal', random_state=SEED)
        else:
            self.scaler = RobustScaler()

        X_scaled = self.scaler.fit_transform(X_train)

        self.pca_gene = PCA(n_components=n_pca_gene, random_state=SEED)
        self.pca_cell = PCA(n_components=n_pca_cell, random_state=SEED)

        gene_pca = self.pca_gene.fit_transform(X_scaled[:, gene_idx])
        cell_pca = self.pca_cell.fit_transform(X_scaled[:, cell_idx])

        var_explained_gene = self.pca_gene.explained_variance_ratio_.sum()
        var_explained_cell = self.pca_cell.explained_variance_ratio_.sum()

        X_final = np.hstack([X_scaled, gene_pca, cell_pca])

        return X_final, {'gene_var': var_explained_gene, 'cell_var': var_explained_cell}

    def transform(self, X_test, gene_idx, cell_idx):
        X_scaled = self.scaler.transform(X_test)
        gene_pca = self.pca_gene.transform(X_scaled[:, gene_idx])
        cell_pca = self.pca_cell.transform(X_scaled[:, cell_idx])
        return np.hstack([X_scaled, gene_pca, cell_pca])


# ============================================================================
# 6. MODEL ARCHITECTURES
# ============================================================================
class MoADataset(Dataset):
    def __init__(self, X, y=None):
        self.X = torch.FloatTensor(X)
        self.y = torch.FloatTensor(y) if y is not None else None

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        if self.y is not None:
            return self.X[idx], self.y[idx]
        return self.X[idx]


class MoADeepModel(nn.Module):
    """Architecture Deep Learning multi-label avec BatchNorm + Dropout."""

    def __init__(self, input_dim, n_targets, hidden_dims=[1024, 512, 256], dropout=0.3):
        super(MoADeepModel, self).__init__()

        layers = []
        prev_dim = input_dim

        for hidden_dim in hidden_dims:
            layers.extend([
                nn.utils.weight_norm(nn.Linear(prev_dim, hidden_dim)),
                nn.BatchNorm1d(hidden_dim),
                nn.LeakyReLU(0.1),
                nn.Dropout(dropout)
            ])
            prev_dim = hidden_dim

        self.hidden_layers = nn.Sequential(*layers)
        self.output_layer = nn.utils.weight_norm(nn.Linear(prev_dim, n_targets))

    def forward(self, x):
        x = self.hidden_layers(x)
        x = self.output_layer(x)
        return x


class FocalLoss(nn.Module):
    """Focal Loss pour gérer le déséquilibre extrême des classes."""

    def __init__(self, alpha=0.25, gamma=2.0):
        super(FocalLoss, self).__init__()
        self.alpha = alpha
        self.gamma = gamma

    def forward(self, inputs, targets):
        bce_loss = F.binary_cross_entropy_with_logits(inputs, targets, reduction='none')
        pt = torch.exp(-bce_loss)
        focal_loss = self.alpha * (1 - pt) ** self.gamma * bce_loss
        return focal_loss.mean()


# ============================================================================
# 7. CROSS-VALIDATION STRATEGY (DRUG LEAKAGE PREVENTION)
# ============================================================================
def create_group_folds(df, groups_col='drug_id', n_folds=5, seed=SEED):
    gkf = GroupKFold(n_splits=n_folds)
    folds = list(gkf.split(df, groups=df[groups_col]))

    for i, (train_idx, val_idx) in enumerate(folds):
        train_drugs = set(df.iloc[train_idx][groups_col])
        val_drugs = set(df.iloc[val_idx][groups_col])
        overlap = train_drugs & val_drugs
        assert len(overlap) == 0, f"⚠️ Leakage détecté au fold {i}: {len(overlap)} drugs partagés"

    print(f"✅ {n_folds} folds créés sans data leakage (vérifié)")
    return folds


# ============================================================================
# 8. TRAINING & EVALUATION
# ============================================================================
def train_neural_network(X_train, y_train, X_val, y_val, config=CONFIG):
    train_dataset = MoADataset(X_train, y_train)
    val_dataset = MoADataset(X_val, y_val)

    train_loader = DataLoader(train_dataset, batch_size=config['batch_size'], shuffle=True)
    val_loader = DataLoader(val_dataset, batch_size=config['batch_size'], shuffle=False)

    model = MoADeepModel(
        input_dim=X_train.shape[1],
        n_targets=y_train.shape[1],
        hidden_dims=config['hidden_dims'],
        dropout=config['dropout']
    ).to(DEVICE)

    optimizer = torch.optim.Adam(
        model.parameters(), lr=config['learning_rate'], weight_decay=config['weight_decay'])
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', factor=0.5, patience=3)

    criterion = FocalLoss(alpha=0.25, gamma=2.0)

    best_val_loss = float('inf')
    patience_counter = 0
    history = {'train_loss': [], 'val_loss': []}
    best_model_state = None

    for epoch in range(config['epochs']):
        model.train()
        train_losses = []
        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(DEVICE), y_batch.to(DEVICE)
            optimizer.zero_grad()
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            loss.backward()
            optimizer.step()
            train_losses.append(loss.item())

        model.eval()
        val_losses = []
        val_preds, val_targets = [], []
        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(DEVICE), y_batch.to(DEVICE)
                outputs = model(X_batch)
                loss = criterion(outputs, y_batch)
                val_losses.append(loss.item())
                val_preds.append(torch.sigmoid(outputs).cpu().numpy())
                val_targets.append(y_batch.cpu().numpy())

        train_loss = np.mean(train_losses)
        val_loss = np.mean(val_losses)
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)

        scheduler.step(val_loss)

        if epoch % 5 == 0:
            print(f"Epoch {epoch:3d} | Train Loss: {train_loss:.5f} | Val Loss: {val_loss:.5f}")

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_model_state = model.state_dict()
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= config['patience']:
                print(f"⏹️ Early stopping à l'epoch {epoch}")
                break

    if best_model_state is not None:
        model.load_state_dict(best_model_state)

    val_preds = np.vstack(val_preds)
    val_targets = np.vstack(val_targets)

    return model, history, val_preds, val_targets


def compute_evaluation_metrics(y_true, y_pred, target_names=None):
    y_pred_clipped = np.clip(y_pred, 1e-6, 1 - 1e-6)
    metrics = {}

    metrics['log_loss_mean'] = np.mean([
        log_loss(y_true[:, i], y_pred_clipped[:, i], labels=[0, 1])
        for i in range(y_true.shape[1])
    ])

    aucs = []
    for i in range(y_true.shape[1]):
        if len(np.unique(y_true[:, i])) > 1:
            aucs.append(roc_auc_score(y_true[:, i], y_pred_clipped[:, i]))
    metrics['auc_mean'] = np.mean(aucs) if aucs else np.nan
    metrics['n_valid_auc'] = len(aucs)

    aps = []
    for i in range(y_true.shape[1]):
        if y_true[:, i].sum() > 0:
            aps.append(average_precision_score(y_true[:, i], y_pred_clipped[:, i]))
    metrics['average_precision_mean'] = np.mean(aps) if aps else np.nan

    return metrics


def train_lightgbm_baseline(X_train, y_train, X_val, y_val, target_cols):
    val_preds = np.zeros((len(X_val), len(target_cols)))

    for i, target in enumerate(target_cols):
        if y_train[:, i].sum() < 2:
            continue
        model = lgb.LGBMClassifier(
            n_estimators=100, num_leaves=31, learning_rate=0.05,
            class_weight='balanced', random_state=SEED, verbosity=-1
        )
        model.fit(X_train, y_train[:, i])
        val_preds[:, i] = model.predict_proba(X_val)[:, 1]

    return val_preds


def cross_validate_pipeline(df, gene_cols, cell_cols, target_cols,
                              n_folds=5, model_type='nn'):
    folds = create_group_folds(df, n_folds=n_folds)

    all_metrics = []
    oof_predictions = np.zeros((len(df), len(target_cols)))
    last_history = None

    for fold_idx, (train_idx, val_idx) in enumerate(folds):
        print(f"\n{'='*50}")
        print(f"FOLD {fold_idx + 1}/{n_folds} [{model_type.upper()}]")
        print(f"{'='*50}")

        train_df = df.iloc[train_idx]
        val_df = df.iloc[val_idx]

        X_train_raw = train_df[gene_cols + cell_cols].values
        y_train = train_df[target_cols].values
        X_val_raw = val_df[gene_cols + cell_cols].values
        y_val = val_df[target_cols].values

        fe = FeatureEngineer(method='rankgauss')
        gene_idx = list(range(len(gene_cols)))
        cell_idx = list(range(len(gene_cols), len(gene_cols) + len(cell_cols)))

        X_train, _ = fe.fit_transform(X_train_raw, gene_idx, cell_idx)
        X_val = fe.transform(X_val_raw, gene_idx, cell_idx)

        if model_type == 'nn':
            model, history, val_preds, val_targets = train_neural_network(
                X_train, y_train, X_val, y_val)
            last_history = history
        elif model_type == 'lgb':
            val_preds = train_lightgbm_baseline(X_train, y_train, X_val, y_val, target_cols)

        oof_predictions[val_idx] = val_preds

        metrics = compute_evaluation_metrics(y_val, val_preds)
        metrics['fold'] = fold_idx
        all_metrics.append(metrics)

        print(f"📊 Fold {fold_idx+1} - Log Loss: {metrics['log_loss_mean']:.5f} | "
              f"AUC: {metrics['auc_mean']:.4f}")

    return pd.DataFrame(all_metrics), oof_predictions, last_history


# ============================================================================
# 9. VISUALISATIONS PUBLICATION-READY
# ============================================================================
def plot_training_curves(history, save_path='figure2_training.png'):
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.plot(history['train_loss'], label='Training Loss', linewidth=2)
    ax.plot(history['val_loss'], label='Validation Loss', linewidth=2)
    ax.set_xlabel('Epoch')
    ax.set_ylabel('Focal Loss')
    ax.set_title('Training and Validation Loss Curves')
    ax.legend()
    ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig(f"{CONFIG['output_dir']}/{save_path}", dpi=300)
    plt.show()


def plot_model_comparison(nn_results, lgb_results, save_path='figure3_comparison.png'):
    nn_results = nn_results.copy()
    lgb_results = lgb_results.copy()
    nn_results['model'] = 'Neural Network'
    lgb_results['model'] = 'LightGBM'
    combined = pd.concat([nn_results, lgb_results], ignore_index=True)

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    sns.boxplot(data=combined, x='model', y='log_loss_mean', ax=axes[0])
    axes[0].set_title('A. Log Loss Comparison Across Models')
    axes[0].set_ylabel('Log Loss (lower is better)')

    sns.boxplot(data=combined, x='model', y='auc_mean', ax=axes[1])
    axes[1].set_title('B. AUC-ROC Comparison Across Models')
    axes[1].set_ylabel('AUC-ROC (higher is better)')

    plt.tight_layout()
    plt.savefig(f"{CONFIG['output_dir']}/{save_path}", dpi=300)
    plt.show()
    return combined


def plot_per_moa_performance(y_true, y_pred, target_cols, top_n=20,
                               save_path='figure4_per_moa.png'):
    aucs = {}
    for i, target in enumerate(target_cols):
        if len(np.unique(y_true[:, i])) > 1:
            aucs[target] = roc_auc_score(y_true[:, i], y_pred[:, i])

    aucs_series = pd.Series(aucs).sort_values(ascending=False)

    fig, axes = plt.subplots(1, 2, figsize=(16, 8))

    aucs_series.head(top_n).plot(kind='barh', ax=axes[0], color='#2ecc71')
    axes[0].set_title(f'A. Top {top_n} Best Predicted MoA')
    axes[0].set_xlabel('AUC-ROC')
    axes[0].invert_yaxis()

    aucs_series.tail(top_n).plot(kind='barh', ax=axes[1], color='#e74c3c')
    axes[1].set_title(f'B. Bottom {top_n} Worst Predicted MoA')
    axes[1].set_xlabel('AUC-ROC')
    axes[1].invert_yaxis()

    plt.tight_layout()
    plt.savefig(f"{CONFIG['output_dir']}/{save_path}", dpi=300)
    plt.show()

    return aucs_series


def statistical_comparison_models(nn_results, lgb_results):
    n = min(len(nn_results), len(lgb_results))
    stat, p_value = wilcoxon(
        nn_results['log_loss_mean'].values[:n],
        lgb_results['log_loss_mean'].values[:n]
    )
    result = pd.DataFrame([{
        'Model_A': 'Neural Network',
        'Model_B': 'LightGBM',
        'Statistic': stat,
        'p_value': p_value,
        'Significant (p<0.05)': p_value < 0.05
    }])
    return result


# ============================================================================
# 10. MAIN EXECUTION PIPELINE
# ============================================================================
def main():
    print("="*70)
    print("MULTI-LABEL DEEP LEARNING FOR MoA PREDICTION")
    print("="*70)

    # --- 1. Chargement des données ---
    loader = MoADataLoader(data_dir='./').load_all()
    train_merged = loader.merge_train_data()
    train_active = train_merged[train_merged['cp_type'] == 'trt_cp'].copy()

    # --- 2. EDA pour l'article ---
    eda = MoAExploratoryAnalysis(loader)

    summary_stats = eda.compute_summary_statistics()
    print("\n📊 TABLE 1 - Dataset Summary Statistics")
    print(summary_stats)
    summary_stats.to_csv(f"{CONFIG['output_dir']}/table1_summary.csv")

    moa_counts, imbalance_stats = eda.analyze_label_imbalance()
    print("\n📊 TABLE 2 - Label Imbalance Statistics")
    print(pd.Series(imbalance_stats))
    pd.Series(imbalance_stats).to_csv(f"{CONFIG['output_dir']}/table2_imbalance.csv")

    drug_counts, leakage_stats = eda.analyze_drug_leakage_risk()
    print("\n📊 TABLE 3 - Drug Replication Statistics")
    print(pd.Series(leakage_stats))
    pd.Series(leakage_stats).to_csv(f"{CONFIG['output_dir']}/table3_leakage.csv")

    eda.generate_figure_1_overview()

    # --- 3. Cross-Validation & Model Training: Neural Network ---
    print("\n" + "="*70)
    print("CROSS-VALIDATION: NEURAL NETWORK MODEL")
    print("="*70)

    nn_results, nn_oof, nn_history = cross_validate_pipeline(
        train_active, loader.gene_cols, loader.cell_cols,
        loader.target_cols, n_folds=CONFIG['n_folds'], model_type='nn'
    )

    print("\n📊 TABLE 4 - Neural Network CV Results")
    print(nn_results)
    print(f"\nMean Log Loss: {nn_results['log_loss_mean'].mean():.5f} "
          f"± {nn_results['log_loss_mean'].std():.5f}")
    nn_results.to_csv(f"{CONFIG['output_dir']}/table4_nn_results.csv")

    if nn_history is not None:
        plot_training_curves(nn_history)

    # --- 4. Baseline Comparison (LightGBM) ---
    print("\n" + "="*70)
    print("CROSS-VALIDATION: LIGHTGBM BASELINE")
    print("="*70)

    top_targets = moa_counts.head(50).index.tolist()

    lgb_results, lgb_oof, _ = cross_validate_pipeline(
        train_active, loader.gene_cols, loader.cell_cols,
        top_targets, n_folds=CONFIG['n_folds'], model_type='lgb'
    )

    print("\n📊 TABLE 5 - LightGBM Baseline Results")
    print(lgb_results)
    lgb_results.to_csv(f"{CONFIG['output_dir']}/table5_lgb_results.csv")

    # --- 5. Comparaison statistique & visualisations finales ---
    comparison_df = plot_model_comparison(nn_results, lgb_results)

    stat_test = statistical_comparison_models(nn_results, lgb_results)
    print("\n📊 Wilcoxon Signed-Rank Test (NN vs LightGBM)")
    print(stat_test)
    stat_test.to_csv(f"{CONFIG['output_dir']}/wilcoxon_test.csv")

    aucs_per_moa = plot_per_moa_performance(
        train_active[loader.target_cols].values,
        nn_oof,
        loader.target_cols
    )

    print("\n" + "="*70)
    print("✅ PIPELINE TERMINÉ - Résultats sauvegardés dans:", CONFIG['output_dir'])
    print("="*70)

    return {
        'loader': loader,
        'nn_results': nn_results,
        'nn_oof': nn_oof,
        'lgb_results': lgb_results,
        'lgb_oof': lgb_oof,
        'summary_stats': summary_stats,
        'imbalance_stats': imbalance_stats,
        'leakage_stats': leakage_stats,
        'stat_test': stat_test,
        'aucs_per_moa': aucs_per_moa
    }


# ============================================================================
# EXÉCUTION
# ============================================================================
results = main()
```
